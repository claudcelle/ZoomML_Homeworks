# Q4. Missing values

## How many columns in the dataset have missing values?

```python

def count_columns_with_missing_values(df:pd.DataFrame)->int:
    bool_series = df.isna().any() # boolean series: True if column has NaN values, False if has not
    print(f"Numbers of columns: {len(df.columns)}")
    print(f"Numbers of columns with NaNs: {bool_series.sum()}")
    print(f"Numbers of columns without NaNs:  {(~bool_series).sum()}")

    return int(bool_series.sum()) # Sum of boolean series: Trues counts as 1s Falses as 0s

count_columns_with_missing_values(df)
```

Q7. Sum of weights

- Select all the cars from Asia
- Select only columns vehicle_weight and model_year
- Select the first 7 values
- Get the underlying NumPy array. Let's call it X.
- Compute matrix-matrix multiplication between the transpose of X and X. To get the transpose, use X.T. Let's call the result XTX.
- Invert XTX.
- Create an array y with values [1100, 1300, 800, 900, 1000, 1100, 1200].
- Multiply the inverse of XTX with the transpose of X, and then multiply the result by y. Call the result w.
- What's the sum of all the elements of the result?



```python
from numpy.linalg import inv,matmul
import numpy as np
```

```python
X = (
    df
    [df['origin'] == "Asia"]            # filter by origin
    [['vehicle_weight','model_year']]   # select columns
    .head(7)                            # select first 7
    .to_numpy()                         # convert to numpy array
    )

XTX = X.T@X
# (
    # X.transpose().dot(X)
# )

# inv(XTX)
```

```python

XTX_inv = inv(XTX)
y = np.array(
    [1100, 1300, 800, 900, 1000, 1100, 1200])


#- Multiply the inverse of XTX with the transpose of X, and then multiply the result by y. Call the result w.
# w = XTX_inv.dot(X.T).dot(y)
w = XTX_inv @ X.T @ y


w.sum()
```
